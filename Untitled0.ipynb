```python
df.groupby('Sex')['Survived'].mean().plot(kind='bar')
plt.title('Survival Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Survival Rate')
plt.show()
```

```python
df.groupby('Pclass')['Survived'].mean().plot(kind='bar')
plt.title('Survival Rate by Passenger Class')
plt.ylabel('Survival Ratio')
plt.show()
```

```python
df['Age'].plot(kind='hist', bins=40)
plt.title('Age Distribution of passengers')
plt.xlabel('Age')
plt.show()
```

```python
print("Mean age:", np.mean(df['Age'].dropna()))
print("Median age:", np.median(df['Age'].dropna()))
```

```python
print("Mean age:", df['Age'].mean())
print("Median age:", df['Age'].median())
```

## Titanic Dataset — Data Exploration Project

In this project, I explored the Titanic dataset using Python, Pandas, NumPy, and Matplotlib.

**Tools used:** Pandas (data loading and analysis), NumPy (numerical calculations), Matplotlib (visualization)

**Dataset:** 891 passengers, 12 columns (PassengerId, Survived, Pclass, Name, Sex, Age, SibSp, Parch, Ticket, Fare, Cabin, Embarked)

**Key findings:**
- Overall survival rate was about 38.4%.
- Female passengers had a survival rate of about 74.2%, while male passengers had about 18.9%.
- 1st class passengers had the highest survival rate at about 63%, followed by 2nd class (47.3%) and 3rd class (24.2%).
- The Age column had 177 missing values and Cabin had 687 missing values, which were handled before analysis.
- Mean age of passengers was about 29.7, median age was about 28.0.
- Most passengers were in the 20-40 age range, based on the age distribution histogram.

**What I learned:**
- How to load and explore a real dataset using Pandas (.head(), .shape, .info(), .describe())
- How to group data and calculate statistics (groupby, mean, value_counts)
- How to check and handle missing values (isnull().sum())
- How to create basic visualizations with Matplotlib (bar charts, histograms)
- Basic use of NumPy for numerical calculations (mean, median), and how Pandas methods do the same thing more simply
